# Library and data import

```python
# Importing our libraries and model. Pandas and numpy are used to deal with data and numerical data. Scikit learn is an ML library that provides algorithms and eval tools. 
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
```

```python
# Loading in data
# We are loading in our provided dataset as a csv. 
data = pd.read_csv("C:/Users/Ryan Brunelle/Desktop/d682-ai-optimization-for-computer-scientists/DQN1 Dataset.csv")
```

```python
# Checking data loaded in
data.head(10)
```

# After verifying the dataset loaded in correctly we can see it contains all features including the historical air quality, weather conditions, and corresponding health risk scores I need to train the model. 

```python
# Pushing to repo. 
```

# Now that the data is loaded in, I must identify and define the input features and target variable by setting X to the pm2.5, no2, co2, temp, humidity features of our dataset while setting the Health Risk Score (target variable) to Y. 

```python
# Setting variables (X=Independent, y=Dependent)
X = data[['pm2.5', 'no2', 'co2', 'temp', 'humidity']]
y = data['healthRiskScore']
```

```python
X.head()
```

```python
y.head()
```

# Next we have to split the data sets (X, y) into two sets. One called train in which is the data the model trains itself on, and test which it evaluates its predictive performance on. This is done using Scikit learns "train_test_split" function. 

```python
# Split data to training and testing sets thru TrainTestSplit
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=28)
```

Before fitting the model to our split datasets, we have to call and initialize the RandomForestRegressor model from scikits libray. 

```python
# Initiate the Random Forest Regressor model
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=28
)
```

We can now fit the model to our X and y datasets using .fit function.

```python
# Fitting the model to train and test sets
rf_model.fit(X_train, y_train)
```

Now that the model is fit we can generate the predictions. The predictions are made using the unseen testing data. 

```python
# Model predicting
y_pred = rf_model.predict(X_test)
```

For our evaluation of the results I have chosen to use Mean Squared Erron (rmse) which measures the magnitude of prediction errors and Mean Absolute Percentage Error (mape) which provides a more interpretable percent based error metric. 

```python
# Eval Metrics (mean squared error, mean absolute % error)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mape = mean_absolute_percentage_error(y_test, y_pred)
```

```python
# Printing results
print("RMSE:", rmse)
```

```python
print("MAPE", mape)
```

The RMSE value of 0.2696 indicates that the model’s predictions closely align with observed health risk scores, with relatively small average errors.
The MAPE value of approximately 2.09% suggests that the model is highly accurate, with predictions deviating only slightly from actual values on average.

# TASK 2

```python
# Storing baseline results from Task 1.
baseline_rmse = rmse
baseline_mape = mape


print("Baseline RMSE:", baseline_rmse)
print("Baseline MAPE:", baseline_mape)
```

# B1 - Optimization Technique #1: RandomSearchCV

RandomizedSearchCV improves model performance by testing different hyperparameter combinations to find a strong overall configuration. 

```python
# Import library
from sklearn.model_selection import RandomizedSearchCV
```

```python
# Configure RandomSearchCV and set parameters
rf_param_dist = {
    "n_estimators": [100, 200, 400, 600],
    "max_depth": [None, 5, 10, 15, 20],
    "min_samples_split": [2, 5, 10, 20],
    "min_samples_leaf": [1, 2, 4, 8],
    "max_features": ["sqrt", "log2", 0.8],
    "bootstrap": [True, False]
}

random_search = RandomizedSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_distributions=rf_param_dist,
    n_iter=25,
    scoring="neg_root_mean_squared_error",
    cv=5,
    n_jobs=-1,
    random_state=28
)
```

```python
# Fit model, define, and print new RandomizedSearch Eval metrics.
random_search.fit(X_train, y_train)

optimized_rf_random = random_search.best_estimator_
y_pred_random = optimized_rf_random.predict(X_test)

rmse_random = np.sqrt(mean_squared_error(y_test, y_pred_random))
mape_random = mean_absolute_percentage_error(y_test, y_pred_random)

print("RandomizedSearch RMSE:", rmse_random)
print("RandomizedSearch MAPE:", mape_random)
```

# B1 - Optimization Technique #2: GridSearchCV refinement

GridSearchCV fine tunes the model by exhaustively testing a small, targeted set of hyperparameter values around the best initial results.

```python
# Import Library
from sklearn.model_selection import GridSearchCV
```

```python
# Define Params and Configure GridSearch
grid_params = {
    "n_estimators": [random_search.best_params_["n_estimators"]],
    "max_depth": [random_search.best_params_["max_depth"]],
    "min_samples_split": [random_search.best_params_["min_samples_split"]],
    "min_samples_leaf": [random_search.best_params_["min_samples_leaf"]],
    "max_features": [random_search.best_params_["max_features"]],
    "bootstrap": [random_search.best_params_["bootstrap"]],
}

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_grid=grid_params,
    scoring="neg_root_mean_squared_error",
    cv=5,
    n_jobs=-1
)
```

```python
# Fit model, define, and print new GridsearchCV Eval metrics.
grid_search.fit(X_train, y_train)

optimized_rf_grid = grid_search.best_estimator_
y_pred_grid = optimized_rf_grid.predict(X_test)

rmse_grid = np.sqrt(mean_squared_error(y_test, y_pred_grid))
mape_grid = mean_absolute_percentage_error(y_test, y_pred_grid)

print("GridSearch RMSE:", rmse_grid)
print("GridSearch MAPE:", mape_grid)
```

# B2 - Regularization Methods - max_depth & min_samples_leaf

Max_depth limits how deep each decision tree can grow.


Min_samples_leaf sets minimum number of training samples required in a node and requires each leaf to represent more samples.

```python
# Define params and fit with new regularization vars. 
regularized_rf = RandomForestRegressor(
    n_estimators=300,
    random_state=28,
    max_depth=10,
    min_samples_leaf=4
)

regularized_rf.fit(X_train, y_train)

# Evaluate regularized model
y_pred_reg = regularized_rf.predict(X_test)
rmse_reg = np.sqrt(mean_squared_error(y_test, y_pred_reg))
mape_reg = mean_absolute_percentage_error(y_test, y_pred_reg)

print("Regularized RF RMSE:", rmse_reg)
print("Regularized RF MAPE:", mape_reg)
```

# B3 - Ensemble Technique #1: ExtraTrees

Extra Trees increase randomness during tree splitting,  improving generalization and reducing variance.

```python
# Import Library
from sklearn.ensemble import ExtraTreesRegressor
```

```python
# Define and establish extra_trees model and params.
extra_trees = ExtraTreesRegressor(
    n_estimators=500,
    random_state=28,
    n_jobs=-1
)
```

```python
# Fit model
extra_trees.fit(X_train, y_train)
```

```python
# Evaluate Extra Trees model and print.
y_pred_et = extra_trees.predict(X_test)
rmse_et = np.sqrt(mean_squared_error(y_test, y_pred_et))
mape_et = mean_absolute_percentage_error(y_test, y_pred_et)

print("ExtraTrees RMSE:", rmse_et)
print("ExtraTrees MAPE:", mape_et)
```

# B3 - Ensemble Technique #2: Gradient Boosting

Gradient Boosting builds models sequentially with each new model correcting errors from previous one.

```python
# Import Library
from sklearn.ensemble import GradientBoostingRegressor
```

```python
# Define GRboost
gbr = GradientBoostingRegressor(random_state=42)
```

```python
# Fit model
gbr.fit(X_train, y_train)
```

```python
# Evaluate Gradient Boosting model and print.
y_pred_gbr = gbr.predict(X_test)
rmse_gbr = np.sqrt(mean_squared_error(y_test, y_pred_gbr))
mape_gbr = mean_absolute_percentage_error(y_test, y_pred_gbr)

print("GradientBoosting RMSE:", rmse_gbr)
print("GradientBoosting MAPE:", mape_gbr)
```
